```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# --- Load the image ---
img = cv2.imread('salt&pepper.PNG')   # Replace with your image file
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# --- Apply Box Filter with normalization (average) ---
box_normalized = cv2.boxFilter(img, ddepth=-1, ksize=(5,5), normalize=True)

# --- Apply Box Filter without normalization (sum only) ---
box_sum = cv2.boxFilter(img, ddepth=-1, ksize=(5,5), normalize=False)

# --- Display results ---
titles = ['Original Image', 'Box Filter (normalize=True)', 'Box Filter (normalize=False)']
images = [img, box_normalized, box_sum]

plt.figure(figsize=(12, 6))
for i in range(3):
    plt.subplot(1, 3, i+1)
    plt.imshow(np.clip(images[i], 0, 255).astype(np.uint8))
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```
