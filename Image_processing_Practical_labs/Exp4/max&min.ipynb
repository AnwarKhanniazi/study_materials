```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# --- Read RGB image ---
img = cv2.imread('cat.jpg')       # Load image
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Convert from BGR to RGB

# --- Define 3x3 kernel ---
kernel = np.ones((3,3), np.uint8)

# --- Apply Max (Dilation) and Min (Erosion) Filters ---
max_filtered = cv2.dilate(img, kernel)   # Max filter
min_filtered = cv2.erode(img, kernel)    # Min filter

# --- Display Results ---
plt.figure(figsize=(12,5))
plt.subplot(1,3,1)
plt.imshow(img)
plt.title('Original RGB Image')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(max_filtered)
plt.title('Max Filter (3x3)')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(min_filtered)
plt.title('Min Filter (3x3)')
plt.axis('off')

plt.show()
```
