```python
import cv2
import numpy as np
import random
import matplotlib.pyplot as plt

# Read the image
image = cv2.imread('pk.jpg')
if image is None:
    raise FileNotFoundError("Image 'pk.jpg' not found.")
    
# Convert to RGB for correct display in matplotlib
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Function to add impulse (salt and pepper) noise
def add_impulse_noise(img, prob):
    output = np.copy(img)
    # Generate random numbers for each pixel
    random_matrix = np.random.rand(*img.shape[:2])
    
    # Salt (white) noise
    output[random_matrix < (prob / 2)] = 255
    
    # Pepper (black) noise
    output[random_matrix > 1 - (prob / 2)] = 0
    
    return output

# Probability of noise
noise_prob = 0.05  # 5% pixels corrupted

# Add noise
noisy_image = add_impulse_noise(image, noise_prob)

# Show images
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title('Original Image')
plt.imshow(image)
plt.axis('off')

plt.subplot(1,2,2)
plt.title('Impulse Noisy Image')
plt.imshow(noisy_image)
plt.axis('off')

plt.show()

# Save noisy image
cv2.imwrite('pk_impulse.jpg', cv2.cvtColor(noisy_image, cv2.COLOR_RGB2BGR))
print("Noisy image saved as 'pk_impulse.jpg'")
```
