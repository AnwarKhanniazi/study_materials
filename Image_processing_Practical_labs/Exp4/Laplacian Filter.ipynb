```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# --- Read and prepare image ---
img = cv2.imread('cat.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

# --- Positive Laplacian Kernel ---
positive_kernel = np.array([[0, -1, 0],
                            [-1, 4, -1],
                            [0, -1, 0]])

# --- Negative Laplacian Kernel ---
negative_kernel = np.array([[0, 1, 0],
                            [1, -4, 1],
                            [0, 1, 0]])

# --- Apply filters ---
lap_pos = cv2.filter2D(gray, -1, positive_kernel)
lap_neg = cv2.filter2D(gray, -1, negative_kernel)

# --- Display results ---
plt.figure(figsize=(12,5))

plt.subplot(1,3,1)
plt.imshow(gray, cmap='gray')
plt.title('Original Grayscale')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(lap_pos, cmap='gray')
plt.title('Positive Laplacian Filter')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(lap_neg, cmap='gray')
plt.title('Negative Laplacian Filter')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# --- Read RGB Image ---
img = cv2.imread('cat.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# --- Define Positive and Negative Laplacian Kernels ---
positive_kernel = np.array([[0, -1, 0],
                            [-1, 4, -1],
                            [0, -1, 0]])

negative_kernel = np.array([[0, 1, 0],
                            [1, -4, 1],
                            [0, 1, 0]])

# --- Apply filters directly on RGB image ---
laplacian_positive = cv2.filter2D(img, -1, positive_kernel)
laplacian_negative = cv2.filter2D(img, -1, negative_kernel)

# --- Display Results ---
plt.figure(figsize=(12,5))

plt.subplot(1,3,1)
plt.imshow(img)
plt.title('Original RGB Image')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(laplacian_positive)
plt.title('Positive Laplacian (RGB)')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(laplacian_negative)
plt.title('Negative Laplacian (RGB)')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# --- Read RGB image ---
img = cv2.imread('cat.jpg')             # Replace with your image path
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# --- Apply Laplacian filter ---
laplacian = cv2.Laplacian(img, cv2.CV_64F)     # High-pass filter
laplacian = cv2.convertScaleAbs(laplacian)     # Convert back to 8-bit

# --- Sharpened image = Original + (alpha * Laplacian) ---
alpha = 1.0  # sharpening strength (try values: 0.5 to 2.0)
sharpened = cv2.addWeighted(img, 1, laplacian, alpha, 0)

# --- Display results ---
plt.figure(figsize=(12,5))

plt.subplot(1,3,1)
plt.imshow(img)
plt.title('Original RGB Image')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(laplacian)
plt.title('Laplacian Filter (Edges)')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(sharpened)
plt.title('Sharpened Image (RGB)')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# --- Step 1: Read image in RGB ---
img = cv2.imread('cat.jpg')              # Replace with your image path
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# --- Step 2: Apply Laplacian filter (edge enhancement internally) ---
laplacian = cv2.Laplacian(img, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)

# --- Step 3: Sharpen the image (add Laplacian to original) ---
# This enhances details but doesn't display edge map
alpha = 1.0   # sharpening strength (0.3 = mild, 1.0 = normal, 1.5 = strong)
sharpened = cv2.addWeighted(img, 1.0, laplacian, alpha, 0)

# --- Step 4: Display results ---
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(img)
plt.title('Original Image')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(sharpened)
plt.title('Sharpened Image (Laplacian)')
plt.axis('off')

plt.show()
```
