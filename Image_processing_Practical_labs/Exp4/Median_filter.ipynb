```python
import cv2
import matplotlib.pyplot as plt

# --- Load image ---
img = cv2.imread('salt&pepper.PNG')      # Replace with your image file
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# --- Apply Median Filter ---
median_filtered = cv2.medianBlur(img, 5)   # Kernel size = 5x5

# --- Display results ---
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(median_filtered)
plt.title("Median Filtered (5x5)")
plt.axis("off")

plt.tight_layout()
plt.show()
```
