```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
image = cv2.imread('pk.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Function to add Salt & Pepper noise
def add_salt_pepper_noise(img, salt_prob=0.02, pepper_prob=0.02):
    noisy = np.copy(img)
    total_pixels = img.size // img.shape[2]

    # Add salt (white) noise
    num_salt = int(total_pixels * salt_prob)
    coords = [np.random.randint(0, i - 1, num_salt) for i in img.shape[:2]]
    noisy[coords[0], coords[1]] = [255, 255, 255]

    # Add pepper (black) noise
    num_pepper = int(total_pixels * pepper_prob)
    coords = [np.random.randint(0, i - 1, num_pepper) for i in img.shape[:2]]
    noisy[coords[0], coords[1]] = [0, 0, 0]

    return noisy

# Apply noise
noisy_image = add_salt_pepper_noise(image)

# Display results
plt.figure(figsize=(10, 6))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(image)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Salt & Pepper Noise: Random occurrence of black and white pixels")
plt.imshow(noisy_image)
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
image = cv2.imread('pk.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Function to add Gaussian noise
def add_gaussian_noise(img, mean=0, var=0.01):
    # Convert image to float for precision
    img = img / 255.0
    sigma = var ** 0.5
    gaussian = np.random.normal(mean, sigma, img.shape)
    
    # Add the Gaussian noise
    noisy = img + gaussian
    
    # Clip the values to [0,1] and convert back to 0–255
    noisy = np.clip(noisy, 0, 1)
    noisy = (noisy * 255).astype(np.uint8)
    return noisy

# Apply Gaussian noise
noisy_image = add_gaussian_noise(image, mean=0, var=0.02)

# Display results
plt.figure(figsize=(10, 6))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(image)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Gaussian Noise: Random variation based on normal distribution")
plt.imshow(noisy_image)
plt.axis('off')

plt.show()

```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
image = cv2.imread('pk.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Function to add Gaussian noise
def add_gaussian_noise(img, mean=0, var=0.01):
    img = img / 255.0
    sigma = var ** 0.5
    gaussian = np.random.normal(mean, sigma, img.shape)
    noisy = img + gaussian
    noisy = np.clip(noisy, 0, 1)
    noisy = (noisy * 255).astype(np.uint8)
    return noisy

# Add Gaussian noise
noisy_image = add_gaussian_noise(image, mean=0, var=0.02)

# Display original and noisy images
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(noisy_image)
plt.title("Image with Gaussian Noise")
plt.axis('off')

# Plot histograms
plt.subplot(2, 2, 3)
plt.title("Original Image Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.hist(image.ravel(), bins=256, color='blue', alpha=0.7)

plt.subplot(2, 2, 4)
plt.title("Gaussian Noise Image Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.hist(noisy_image.ravel(), bins=256, color='red', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Mean and standard deviation for Gaussian distribution
mu = 0        # mean
sigma = 1     # standard deviation

# Generate values from -5σ to +5σ
x = np.linspace(mu - 5*sigma, mu + 5*sigma, 1000)

# Gaussian PDF formula
P = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-((x - mu) ** 2) / (2 * sigma ** 2))

# Plot the Gaussian curve
plt.figure(figsize=(8, 5))
plt.plot(x, P, color='blue', linewidth=2)
plt.title('Gaussian (Normal) Distribution PDF')
plt.xlabel('x (Noise Value)')
plt.ylabel('Probability Density')
plt.grid(True)
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read your image (pk.jpg)
image = cv2.imread('pk.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Function to add Gaussian noise
def add_gaussian_noise(img, mean=0, var=0.01):
    img = img / 255.0
    sigma = var ** 0.5
    gaussian = np.random.normal(mean, sigma, img.shape)
    noisy = img + gaussian
    noisy = np.clip(noisy, 0, 1)
    noisy = (noisy * 255).astype(np.uint8)
    return noisy

# Add Gaussian noise
noisy_image = add_gaussian_noise(image, mean=0, var=0.02)

# Save the noisy image
cv2.imwrite('pj.jpg', cv2.cvtColor(noisy_image, cv2.COLOR_RGB2BGR))
print("✅ Gaussian noisy image saved as pj.jpg")

# Generate Gaussian PDF for visualization
mu = 0
sigma = 1
x = np.linspace(mu - 5*sigma, mu + 5*sigma, 1000)
P = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-((x - mu)**2) / (2 * sigma**2))

# Plot everything
plt.figure(figsize=(14, 10))

# 1️⃣ Original Image
plt.subplot(2, 2, 1)
plt.imshow(image)
plt.title("Original Image (pk.jpg)")
plt.axis('off')

# 2️⃣ Gaussian Noisy Image
plt.subplot(2, 2, 2)
plt.imshow(noisy_image)
plt.title("Gaussian Noisy Image (pj.jpg)")
plt.axis('off')

# 3️⃣ Histograms
plt.subplot(2, 2, 3)
plt.hist(image.ravel(), bins=256, color='blue', alpha=0.6, label='Original')
plt.hist(noisy_image.ravel(), bins=256, color='red', alpha=0.6, label='Noisy')
plt.title("Histograms of Original and Noisy Images")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.legend()

# 4️⃣ Gaussian PDF Curve
plt.subplot(2, 2, 4)
plt.plot(x, P, color='green', linewidth=2)
plt.title("Gaussian Probability Density Function (PDF)")
plt.xlabel("x (Noise Value)")
plt.ylabel("Probability Density")
plt.grid(True)

plt.tight_layout()
plt.show()
```
