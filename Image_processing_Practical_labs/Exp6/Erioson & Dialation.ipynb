```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# --- 1️⃣ Read image ---
img = cv2.imread('pk.jpg', cv2.IMREAD_GRAYSCALE)
if img is None:
    raise FileNotFoundError("❌ Image not found. Please make sure 'pk.jpg' exists in this folder.")

# --- 2️⃣ Define Structuring Element (Kernel) ---
kernel_size = 5  # You can change it (3, 5, 7, ...)
kernel = np.ones((kernel_size, kernel_size), np.uint8)

# --- 3️⃣ Apply Erosion and Dilation ---
erosion = cv2.erode(img, kernel, iterations=1)
dilation = cv2.dilate(img, kernel, iterations=1)

# --- 4️⃣ Display Results ---
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(erosion, cmap='gray')
plt.title(f'Erosion (kernel={kernel_size}x{kernel_size})')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(dilation, cmap='gray')
plt.title(f'Dilation (kernel={kernel_size}x{kernel_size})')
plt.axis('off')

plt.tight_layout()
plt.show()
```
