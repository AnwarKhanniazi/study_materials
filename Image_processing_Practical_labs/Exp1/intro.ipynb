### 1. Importing Libraries
We will use the following libraries:
- OpenCV for image and video processing.
- NumPy for handling image arrays and performing mathematical operations.
- Matplotlib for visualizing images and data.
- PyTorch for image classification with a pre-trained deep network (later).

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### 2. Working with Images in OpenCV

```python
# Reading an image using OpenCV
image = cv2.imread('pk.jpg')  # Replace 'sample_image.jpg' with your image path

# Convert the image from BGR to RGB (OpenCV uses BGR by default)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Displaying the image using Matplotlib
plt.imshow(image_rgb)
plt.title('Original Image')
plt.axis('off')  # Turn off axis labels
plt.show()
```

### 3. Working with Videos in OpenCV

```python
# Capture video from a file or webcam
video = cv2.VideoCapture('sample_video.m4v')  # Replace with 0 to use webcam

while True:
    ret, frame = video.read()  # Capture frame-by-frame
    if not ret:
        break  # Exit if video ends
    
    cv2.imshow('Video Frame', frame)  # Show each frame in a window
    
    if cv2.waitKey(25) & 0xFF == ord('q'):
        break  # Press 'q' to quit the video window

video.release()
cv2.destroyAllWindows()
```

### 4. Experimenting with Image Channels (RGB)

```python
# Splitting the image into its R, G, B channels
B, G, R = cv2.split(image)

# Displaying individual channels
plt.figure(figsize=(20, 10))
plt.subplot(1, 3, 1)
plt.imshow(R, cmap='gray')
plt.title('Red Channel')

plt.subplot(1, 3, 2)
plt.imshow(G, cmap='gray')
plt.title('Green Channel')

plt.subplot(1, 3, 3)
plt.imshow(B, cmap='gray')
plt.title('Blue Channel')

plt.show()
```

### 5. Image Blurring and Sharpening

```python
# Applying Gaussian Blur (Smoothing)
blurred_image = cv2.GaussianBlur(image_rgb, (15, 15), 0)

# Creating a kernel for image sharpening
sharpening_kernel = np.array([[0, -1, 0],
                              [-1, 5, -1],
                              [0, -1, 0]])

# Applying the kernel to the image
sharpened_image = cv2.filter2D(blurred_image, -1, sharpening_kernel)
sharpened_image = cv2.filter2D(sharpened_image, -1, sharpening_kernel) #sharpen the result one more time

# Displaying the original, blurred, and sharpened images
plt.figure(figsize=(20, 10))
plt.subplot(1, 3, 1)
plt.imshow(image_rgb)
plt.title('Original Image')

plt.subplot(1, 3, 2)
plt.imshow(blurred_image)
plt.title('Blurred Image')

plt.subplot(1, 3, 3)
plt.imshow(sharpened_image)
plt.title('Sharpened Image')

plt.show()
```

### 6. Image Histograms and Contrast Enhancement

```python
import cv2
import matplotlib.pyplot as plt

# --- 1️⃣ Read the image safely ---
image_path = "pk.jpg"  # change this to your image name
image_bgr = cv2.imread(image_path)

if image_bgr is None:
    raise FileNotFoundError(f"❌ Image '{image_path}' not found!")

# Convert BGR → RGB for display
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

# --- 2️⃣ Display the RGB Histogram ---
plt.figure(figsize=(10, 5))
plt.hist(image_rgb[..., 0].ravel(), bins=256, color='red', alpha=0.5, label='Red')
plt.hist(image_rgb[..., 1].ravel(), bins=256, color='green', alpha=0.5, label='Green')
plt.hist(image_rgb[..., 2].ravel(), bins=256, color='blue', alpha=0.5, label='Blue')
plt.title('RGB Image Histogram (Before Enhancement)')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')
plt.legend()
plt.show()

# --- 3️⃣ Convert to Grayscale ---
gray_image = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2GRAY)

# --- 4️⃣ Apply Histogram Equalization ---
equalized_image = cv2.equalizeHist(gray_image)

# --- 5️⃣ Display Original & Equalized Images ---
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(gray_image, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(equalized_image, cmap='gray')
plt.title('Histogram Equalized Image')
plt.axis('off')

plt.tight_layout()
plt.show()

# --- 6️⃣ Display Histogram Comparison ---
plt.figure(figsize=(10, 5))
plt.hist(gray_image.ravel(), bins=256, color='gray', alpha=0.6, label='Original')
plt.hist(equalized_image.ravel(), bins=256, color='green', alpha=0.6, label='Equalized')
plt.title('Grayscale Histogram Before & After Equalization')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```
