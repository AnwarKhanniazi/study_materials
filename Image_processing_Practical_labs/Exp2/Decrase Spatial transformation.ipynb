```python
import cv2
import matplotlib.pyplot as plt

# --- Load the image ---
img_path = "Images\pk.jpg"  # Change to your image path
img = cv2.imread(img_path)
if img is None:
    raise FileNotFoundError(f"Image not found: {img_path}")

# Convert from BGR → RGB for correct color display
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# --- Resize the image to desired resolutions ---
sizes = [512, 256, 128, 64]  # target resolutions (square images)
resized_images = []

for s in sizes:
    # Resize using INTER_AREA (best for shrinking)
    resized = cv2.resize(img_rgb, (s, s), interpolation=cv2.INTER_AREA)
    resized_images.append((s, resized))

# --- Display all images ---
plt.figure(figsize=(12, 8))
plt.subplot(2, 2, 1)
plt.imshow(resized_images[0][1])
plt.title("512 x 512")
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(resized_images[1][1])
plt.title("256 x 256")
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(resized_images[2][1])
plt.title("128 x 128")
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(resized_images[3][1])
plt.title("64 x 64")
plt.axis('off')

plt.suptitle("Decreased Spatial Resolution", fontsize=14)
plt.tight_layout()
plt.show()
```
