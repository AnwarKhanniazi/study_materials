```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread('eyes.jpg')     # replace with your image filename
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Convert from BGR to RGB for correct display

# Show original image
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title('Original Image')
plt.axis('off')

# Crop image from start (top-left corner)
# Syntax: cropped = img[row_start:row_end, col_start:col_end]
cropped = img[0:150, 0:200]  # first 150 rows and 200 columns

# Show cropped image
plt.subplot(1,2,2)
plt.imshow(cropped)
plt.title('Cropped (Top-Left Corner)')
plt.axis('off')

plt.show()
```
