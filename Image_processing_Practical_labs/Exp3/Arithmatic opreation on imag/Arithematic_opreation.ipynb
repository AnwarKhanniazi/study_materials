```python
import cv2
import matplotlib.pyplot as plt

# Load two images
image1 = cv2.imread('First_img.PNG')
image2 = cv2.imread('Second_img.PNG')

# Convert from BGR to RGB for correct display with matplotlib
image1 = cv2.cvtColor(image1, cv2.COLOR_BGR2RGB)
image2 = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)

# Resize both images to same size (important)
image2 = cv2.resize(image2, (image1.shape[1], image1.shape[0]))

# Perform pixel-wise addition
added = cv2.add(image1, image2)

# Display results
plt.figure(figsize=(10, 10))

plt.subplot(1, 3, 1)
plt.imshow(image1)
plt.title('First Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(image2)
plt.title('Second Image')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(added)
plt.title('Added Image (image1 + image2)')
plt.axis('off')

plt.show()
```

```python
import cv2

# Loading the Images in RGB format
image1 = cv2.imread('First_img.PNG')       # Replace with your filename
image2 = cv2.imread('Second_img.PNG')      # Replace with your filename

# Convert from BGR (default OpenCV format) to RGB
image1 = cv2.cvtColor(image1, cv2.COLOR_BGR2RGB)
image2 = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)

# Print their shapes
print('Image 1 Shape:', image1.shape)
print('Image 2 Shape:', image2.shape)
```

```python
# Performing Weighted Addition & Displaying Results
res = cv2.addWeighted(image1, 0.6, image2, 0.4, 3)

fig = plt.figure(figsize=(10,10))
ax1 = fig.add_subplot(131)
ax2 = fig.add_subplot(132)
ax3 = fig.add_subplot(133)

ax1.imshow(image1, cmap='gray')
ax1.title.set_text('First Image (Weight: 60%)')
ax2.imshow(image2, cmap='gray')
ax2.title.set_text('Second Image (Weight: 40%)')
ax3.imshow(res, cmap='gray')
ax3.title.set_text('Resulting Image after Addition')
plt.show()
```

```python
# Performing Subtraction & Displaying Results
res = cv2.subtract(image1, image2)

fig = plt.figure(figsize=(10,10))
ax1 = fig.add_subplot(131)
ax2 = fig.add_subplot(132)
ax3 = fig.add_subplot(133)

ax1.imshow(image1, cmap='gray')
ax1.title.set_text('First Image')
ax2.imshow(image2, cmap='gray')
ax2.title.set_text('Second Image')
ax3.imshow(res, cmap='gray')
ax3.title.set_text('Resulting Image after Subtraction')
plt.show()
```

```python
# Performing Subtraction & Displaying Results
res = cv2.multiply(image1, image2)

fig = plt.figure(figsize=(10,10))
ax1 = fig.add_subplot(131)
ax2 = fig.add_subplot(132)
ax3 = fig.add_subplot(133)

ax1.imshow(image1, cmap='gray')
ax1.title.set_text('First Image')
ax2.imshow(image2, cmap='gray')
ax2.title.set_text('Second Image')
ax3.imshow(res, cmap='gray')
ax3.title.set_text('Resulting Image after multiplication')
plt.show()
```

```python
# Performing Subtraction & Displaying Results
res = cv2.divide(image1, image2)

fig = plt.figure(figsize=(10,10))
ax1 = fig.add_subplot(131)
ax2 = fig.add_subplot(132)
ax3 = fig.add_subplot(133)

ax1.imshow(image1, cmap='gray')
ax1.title.set_text('First Image')
ax2.imshow(image2, cmap='gray')
ax2.title.set_text('Second Image')
ax3.imshow(res, cmap='gray')
ax3.title.set_text('Resulting Image after divide')
plt.show()
```
